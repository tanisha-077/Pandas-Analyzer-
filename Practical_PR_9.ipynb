```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
class SalesDataAnalyzer:

    def __init__(self, file_path=None):
        self.data = pd.DataFrame()
        self.fig = None

        if file_path:
            self.load_data(file_path)

    def __del__(self):
        pass


    def load_data(self, file_path):
        try:
            self.data = pd.read_csv(file_path)

            if self.data.empty:
                print("Dataset is empty!")
            else:
                print("Dataset loaded successfully!")
                print(self.data.head())

        except Exception as e:
            print("Error loading dataset:", e)


    def explore_data(self):
        if self.data.empty:
            print("Please load a dataset first!")
            return

        print("\n1. First 5 rows")
        print("2. Last 5 rows")
        print("3. Column names")
        print("4. Data types")
        print("5. Basic information")
        print("6. NumPy array indexing and slicing")

        choice = input("Enter your choice: ")

        if choice == "1":
            print(self.data.head())

        elif choice == "2":
            print(self.data.tail())

        elif choice == "3":
            print(self.data.columns.tolist())

        elif choice == "4":
            print(self.data.dtypes)

        elif choice == "5":
            self.data.info()
            print(self.data.describe(include="all"))

        elif choice == "6":
            if "Sales" in self.data.columns:
                arr = self.data["Sales"].to_numpy()
                print("NumPy array:", arr)
                print("First element:", arr[0])
                print("First five elements:", arr[:5])
            else:
                print("Sales column not found!")

        else:
            print("Invalid choice!")

    def clean_data(self):
        if self.data.empty:
            print("Please load a dataset first!")
            return

        print("\n1. Display missing values")
        print("2. Fill missing numeric values with mean")
        print("3. Drop rows with missing values")
        print("4. Replace missing values with a specific value")

        choice = input("Enter your choice: ")

        if choice == "1":
            print(self.data.isnull().sum())

        elif choice == "2":
            columns = self.data.select_dtypes(
                include=np.number
            ).columns

            self.data[columns] = self.data[columns].fillna(
                self.data[columns].mean()
            )
            print("Missing numeric values filled with mean.")

        elif choice == "3":
            self.data.dropna(inplace=True)
            print("Rows with missing values removed.")

        elif choice == "4":
            value = input("Enter replacement value: ")
            self.data.fillna(value, inplace=True)
            print("Missing values replaced.")

        else:
            print("Invalid choice!")

    def mathematical_operations(self):
        if self.data.empty:
            print("Please load a dataset first!")
            return

        if "Sales" not in self.data.columns:
            print("Sales column not found!")
            return

        arr = self.data["Sales"].to_numpy()

        print("Original Sales:", arr)
        print("Sales + 10:", arr + 10)
        print("Sales - 10:", arr - 10)
        print("Sales * 2:", arr * 2)
        print("Sales / 2:", arr / 2)

        if len(arr) >= 2:
            print("First two elements:", arr[:2])
            print("First element:", arr[0])

        if "Profit" in self.data.columns:
            print(
                "Total Sales + Profit:",
                self.data["Sales"] + self.data["Profit"]
            )


    def combine_data(self, other_dataframe=None):
        if self.data.empty:
            print("Please load a dataset first!")
            return

        if other_dataframe is None:
            path = input("Enter another CSV file path: ")

            try:
                other_dataframe = pd.read_csv(path)
            except Exception as e:
                print("Error loading second dataset:", e)
                return

        print("\n1. Concatenate DataFrames")
        print("2. Merge DataFrames using common columns")

        choice = input("Enter your choice: ")

        try:
            if choice == "1":
                result = pd.concat(
                    [self.data, other_dataframe],
                    ignore_index=True
                )
                print(result)

            elif choice == "2":
                common = list(
                    set(self.data.columns) &
                    set(other_dataframe.columns)
                )

                if not common:
                    print("No common columns found!")
                    return

                print("Common columns:", common)
                column = input("Enter common column name: ")

                if column not in common:
                    print("Invalid column!")
                    return

                result = pd.merge(
                    self.data,
                    other_dataframe,
                    on=column
                )
                print(result)

            else:
                print("Invalid choice!")
                return

            save = input("Save combined data? (y/n): ")

            if save.lower() == "y":
                result.to_csv("combined_data.csv", index=False)
                print("Saved as combined_data.csv")

        except Exception as e:
            print("Error combining data:", e)

   
    def split_data(self):
        if self.data.empty:
            print("Please load a dataset first!")
            return

        print("Available columns:", self.data.columns.tolist())
        column = input("Enter column to split by: ")

        if column not in self.data.columns:
            print("Column not found!")
            return

        for value, group in self.data.groupby(column):
            print("\n", value)
            print(group)

            filename = (
                "split_" + str(value).replace(" ", "_") + ".csv"
            )
            group.to_csv(filename, index=False)

        print("Split files saved successfully!")

   
    def search_sort_filter(self):
        if self.data.empty:
            print("Please load a dataset first!")
            return

        print("\n1. Search for a value")
        print("2. Sort data")
        print("3. Filter data")

        choice = input("Enter your choice: ")
        column = input("Enter column name: ")

        if column not in self.data.columns:
            print("Column not found!")
            return

        if choice == "1":
            value = input("Enter value to search: ")
            result = self.data[
                self.data[column].astype(str) == value
            ]
            print(result)

        elif choice == "2":
            order = input("Ascending order? (y/n): ")
            result = self.data.sort_values(
                by=column,
                ascending=(order.lower() == "y")
            )
            print(result)

        elif choice == "3":
            if pd.api.types.is_numeric_dtype(self.data[column]):
                try:
                    value = float(input("Enter minimum value: "))
                    result = self.data[self.data[column] >= value]
                    print(result)
                except ValueError:
                    print("Enter a valid number!")
            else:
                value = input("Enter value to filter: ")
                result = self.data[
                    self.data[column].astype(str) == value
                ]
                print(result)

        else:
            print("Invalid choice!")


    def aggregate_functions(self):
        if self.data.empty:
            print("Please load a dataset first!")
            return

        columns = self.data.select_dtypes(
            include=np.number
        ).columns

        if len(columns) == 0:
            print("No numeric columns found!")
            return

        print("Numeric columns:", columns.tolist())
        column = input("Enter column name: ")

        if column not in columns:
            print("Invalid numeric column!")
            return

        print("Sum:", self.data[column].sum())
        print("Mean:", self.data[column].mean())
        print("Count:", self.data[column].count())
        print("Minimum:", self.data[column].min())
        print("Maximum:", self.data[column].max())

        print("\nGroup-wise total sales, if available:")

        if "Region" in self.data.columns and "Sales" in self.data.columns:
            print(
                self.data.groupby("Region")["Sales"].transform("sum")
            )


    def statistical_analysis(self):
        if self.data.empty:
            print("Please load a dataset first!")
            return

        columns = self.data.select_dtypes(
            include=np.number
        )

        print("\nDescriptive statistics:")
        print(columns.describe())

        print("\nStandard deviation:")
        print(columns.std())

        print("\nVariance:")
        print(columns.var())

        print("\n25th, 50th and 75th percentiles:")
        print(columns.quantile([0.25, 0.50, 0.75]))


    def create_pivot_table(self):
        if self.data.empty:
            print("Please load a dataset first!")
            return

        if not all(
            c in self.data.columns
            for c in ["Region", "Product", "Sales"]
        ):
            print("Region, Product and Sales columns are required!")
            return

        table = pd.pivot_table(
            self.data,
            index="Region",
            columns="Product",
            values="Sales",
            aggfunc="sum",
            fill_value=0
        )

        print(table)

        table.to_csv("pivot_table.csv")
        print("Pivot table saved as pivot_table.csv")

    
    def visualize_data(self):
        if self.data.empty:
            print("Please load a dataset first!")
            return

        print("\n1. Bar Plot")
        print("2. Line Plot")
        print("3. Scatter Plot")
        print("4. Pie Chart")
        print("5. Histogram")
        print("6. Stack Plot")
        print("7. Seaborn Heatmap")
        print("8. Seaborn Box Plot")
        print("9. Subplots")

        choice = input("Enter your choice: ")

        try:
            self.fig, ax = plt.subplots(figsize=(8, 5))

            if choice == "1":
                column = input("Enter category column: ")
                value = input("Enter numeric column: ")

                if column not in self.data.columns or value not in self.data.columns:
                    print("Invalid column!")
                    plt.close(self.fig)
                    return

                grouped = self.data.groupby(column)[value].sum()
                grouped.plot(kind="bar", ax=ax)
                ax.set_title("Bar Plot")

            elif choice == "2":
                column = input("Enter date or X-axis column: ")
                value = input("Enter Y-axis numeric column: ")

                if column not in self.data.columns or value not in self.data.columns:
                    print("Invalid column!")
                    plt.close(self.fig)
                    return

                self.data.plot(
                    x=column, y=value, kind="line", marker="o", ax=ax
                )
                ax.set_title("Line Plot")

            elif choice == "3":
                x = input("Enter X-axis numeric column: ")
                y = input("Enter Y-axis numeric column: ")

                if x not in self.data.columns or y not in self.data.columns:
                    print("Invalid column!")
                    plt.close(self.fig)
                    return

                if not pd.api.types.is_numeric_dtype(self.data[x]) or not pd.api.types.is_numeric_dtype(self.data[y]):
                    print("Both columns must be numeric!")
                    plt.close(self.fig)
                    return

                ax.scatter(self.data[x], self.data[y])
                ax.set_xlabel(x)
                ax.set_ylabel(y)
                ax.set_title("Scatter Plot")

            elif choice == "4":
                column = input("Enter category column: ")
                value = input("Enter numeric column: ")

                if column not in self.data.columns or value not in self.data.columns:
                    print("Invalid column!")
                    plt.close(self.fig)
                    return

                grouped = self.data.groupby(column)[value].sum()
                grouped.plot(kind="pie", autopct="%1.1f%%", ax=ax)
                ax.set_ylabel("")
                ax.set_title("Pie Chart")

            elif choice == "5":
                column = input("Enter numeric column: ")

                if column not in self.data.columns or not pd.api.types.is_numeric_dtype(self.data[column]):
                    print("Invalid numeric column!")
                    plt.close(self.fig)
                    return

                ax.hist(self.data[column].dropna(), bins=5)
                ax.set_xlabel(column)
                ax.set_title("Histogram")

            elif choice == "6":
                if not all(
                    c in self.data.columns
                    for c in ["Product", "Sales", "Profit"]
                ):
                    print("Product, Sales and Profit columns are required!")
                    plt.close(self.fig)
                    return

                grouped = self.data.groupby("Product")[
                    ["Sales", "Profit"]
                ].sum()

                ax.stackplot(
                    range(len(grouped)),
                    grouped["Sales"],
                    grouped["Profit"],
                    labels=["Sales", "Profit"]
                )
                ax.set_xticks(range(len(grouped)))
                ax.set_xticklabels(grouped.index, rotation=45)
                ax.legend()
                ax.set_title("Stack Plot")

            elif choice == "7":
                plt.close(self.fig)
                numeric = self.data.select_dtypes(include=np.number)

                if numeric.empty:
                    print("No numeric columns available!")
                    return

                self.fig, ax = plt.subplots(figsize=(8, 5))
                sns.heatmap(
                    numeric.corr(),
                    annot=True,
                    cmap="coolwarm",
                    ax=ax
                )
                ax.set_title("Correlation Heatmap")

            elif choice == "8":
                column = input("Enter numeric column: ")

                if column not in self.data.columns or not pd.api.types.is_numeric_dtype(self.data[column]):
                    print("Invalid numeric column!")
                    plt.close(self.fig)
                    return

                sns.boxplot(data=self.data, y=column, ax=ax)
                ax.set_title("Box Plot")

            elif choice == "9":
                plt.close(self.fig)

                if not all(
                    c in self.data.columns
                    for c in ["Product", "Sales", "Profit"]
                ):
                    print("Product, Sales and Profit columns are required!")
                    return

                self.fig, axes = plt.subplots(1, 2, figsize=(12, 5))

                self.data.groupby("Product")["Sales"].sum().plot(
                    kind="bar", ax=axes[0], title="Product Sales"
                )

                axes[1].scatter(
                    self.data["Sales"], self.data["Profit"]
                )
                axes[1].set_title("Sales vs Profit")
                axes[1].set_xlabel("Sales")
                axes[1].set_ylabel("Profit")

            else:
                print("Invalid choice!")
                plt.close(self.fig)
                return

            self.fig.tight_layout()
            plt.show()
            print("Visualization displayed successfully!")

        except Exception as e:
            plt.close("all")
            print("Visualization error:", e)
            self.fig = None

   
    def save_visualization(self):
        if self.fig is None:
            print("Create a visualization first!")
            return

        filename = input(
            "Enter file name (e.g. sales_plot.png): "
        )

        if not filename.lower().endswith((".png", ".jpg", ".jpeg")):
            print("Use .png, .jpg or .jpeg extension!")
            return

        try:
            self.fig.savefig(filename, bbox_inches="tight")
            print("Visualization saved as", filename)
        except Exception as e:
            print("Error saving visualization:", e)

   
    def menu(self):
        while True:
            print("\n Sales Data Analysis & Visualization ")
            print("1. Load Dataset")
            print("2. Explore Data")
            print("3. Mathematical Operations")
            print("4. Handle Missing Data")
            print("5. Generate Descriptive Statistics")
            print("6. Data Visualization")
            print("7. Save Visualization")
            print("8. Search, Sort and Filter")
            print("9. Aggregate Functions")
            print("10. Combine DataFrames")
            print("11. Split DataFrame")
            print("12. Create Pivot Table")
            print("13. Exit")

            choice = input("Enter your choice: ")

            if choice == "1":
                path = input("Enter CSV file path: ")
                self.load_data(path)

            elif choice == "2":
                self.explore_data()

            elif choice == "3":
                self.mathematical_operations()

            elif choice == "4":
                self.clean_data()

            elif choice == "5":
                self.statistical_analysis()

            elif choice == "6":
                self.visualize_data()

            elif choice == "7":
                self.save_visualization()

            elif choice == "8":
                self.search_sort_filter()

            elif choice == "9":
                self.aggregate_functions()

            elif choice == "10":
                self.combine_data()

            elif choice == "11":
                self.split_data()

            elif choice == "12":
                self.create_pivot_table()

            elif choice == "13":
                print("Exiting the program. Goodbye!")
                break

            else:
                print("Invalid choice! Try again.")

```

```python
if __name__ == "__main__":
    analyzer = SalesDataAnalyzer("sales_data.csv")
    analyzer.menu()
```
